```python
count = metadata['totalCount']//5
```

```python
next_url = links[1]['href']
```

```python
import time

for i in range(1,count):
    
    url = "https://wft-geo-db.p.rapidapi.com"
    page = url + next_url

    response = requests.get(page, headers=headers)
    print(response)
    
    links = response.json()['links']
    print(links)
    
    for item in links:
        if item['rel']=='next':
            next_url=item['href']
    
    data.extend(response.json()['data'])
    
    time.sleep(1)
    
```

```python
limit = 10
offset = 0
all_data = []

while offset<101:
    url = f"https://wft-geo-db.p.rapidapi.com/v1/geo/cities?countryIds=IN&limit={limit}&offset={offset}"

    response = requests.get(url, headers=headers)
    data_json = response.json()

    cities = data_json.get("data", [])
    print(cities)

    if not cities:
        break

    all_data.extend(cities)
    offset += limit
    print('all_data :',all_data)

    print(f"Fetched {offset} rows")
    time.sleep(1)
```

```python
df=pd.DataFrame(all_data)
```
