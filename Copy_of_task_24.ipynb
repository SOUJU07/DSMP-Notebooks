### **Problem-1:** Make a subplots which have 2 plots.

- For the first chart, draw a scatter plot "*Monitored Cap.(MW)*" vs "*Total Cap. Under Maintenance (MW)*" of top 5 most frequent power stations. Then draw the lines which indicate the average values of these two columns. Change the colors according to the names of the Power Stations.
- For the second chart, draw a scatter plot "*Monitored Cap. (MW)*" vs "*Actual(MU)*" of the top 5 most frequent power stations. Also draw the lines which indicates the average values of these two columns. Change the colors according to the names of the Power Stations.

````python
from matplotlib.lines import lineStyles
# code here
df=pd.read_csv('/content/PowerGeneration - PowerGeneration.csv')
most_freq=df['Power Station'].value_counts().head(5)
new_df=df[df['Power Station'].isin(most_freq.index)]

avg_monitored=new_df['Monitored Cap.(MW)'].mean()
avg_total_cap=new_df['Total Cap. Under Maintenace (MW)'].mean()
avg_actual=new_df['Actual(MU)'].mean()

arr=new_df['Power Station'].unique()

fig,(ax1,ax2) = plt.subplots(1,2,figsize=(17,7))

for i in arr:
  temp_df=new_df[new_df['Power Station']==i]

  ax1.scatter(temp_df['Monitored Cap.(MW)'],
                temp_df['Total Cap. Under Maintenace (MW)'],
                label=i, alpha=0.7)

  ax2.scatter(temp_df['Monitored Cap.(MW)'],
                temp_df['Actual(MU)'],
                label=i, alpha=0.7)

ax1.set_xlabel('Monitored Cap.(MW)')
ax1.set_xlabel('Total Cap. Under Maintenance (MW)')
ax1.set_title('Top 5 Power Stations: Monitored vs Maintenance')
ax1.axvline(avg_monitored, color='red', linestyle='--', label='Avg Monitored')
ax1.axhline(avg_total_cap, color='green', linestyle='--',label='Total Cap')
ax1.legend(loc='upper left', bbox_to_anchor=(0, 1))

ax2.set_xlabel('Monitored Cap.(MW)')
ax2.set_xlabel('Actual MU')
ax2.set_title('Top 5 Power Stations: Monitored vs Actual')
ax2.axvline(avg_monitored, color='red', linestyle='--', label='Avg Monitored')
ax2.axhline(avg_actual, color='green', linestyle='--',label='Avg Actual')
ax2.legend(loc='upper left', bbox_to_anchor=(0, 1))
plt.show()

````

### **Problem-3:** Make a 3D *Surface* plot of this below mathematical equation.

$$z = |x| - |y|$$

````python
# code here

x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

X,Y = np.meshgrid(x,y)

z=np.abs(X) - np.abs(Y)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(X,Y,z,cmap='viridis')
fig.colorbar(p)
````

### **Problem-4:** Draw the 3D *Contour plot* of this below equation:

### `Problem 9:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = sin(\sqrt{x^2 + y^2})$$


````python
# code here
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = np.sin(np.sqrt(xx**2 + yy**2))
fig = plt.figure()
ax = plt.subplot(projection='3d')

ax.plot_surface(xx,yy,z,cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
````

````python
fig = plt.figure()
ax = plt.subplot()

ax.contour(xx,yy,z,cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')

plt.show()
````

````python
fig = plt.figure()
ax = plt.subplot()

ax.contourf(xx,yy,z,cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')

plt.show()
````

### `Problem 10:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = tan(\log_2({x^2 + y^2})$$

````python
# code here
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = np.tan(np.log2(xx**2 + yy**2))
````

````python
fig = plt.figure()
ax = plt.subplot(projection='3d')

ax.plot_surface(xx,yy,z,cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
````

````python
fig = plt.figure()
ax = plt.subplot()

ax.contour(xx,yy,z,cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')

plt.show()
````

````python
fig = plt.figure()
ax = plt.subplot()

ax.contourf(xx,yy,z,cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')

plt.show()
````
